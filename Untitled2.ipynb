```python
from pathlib import Path

# Patch old RDT usage to sklearn OneHotEncoder
path = Path("/content/iterative-dp/Util/gan/ctgan/transformer.py")
text = path.read_text()

text = text.replace(
    "from rdt.transformers import OneHotEncodingTransformer",
    "from sklearn.preprocessing import OneHotEncoder"
)
text = text.replace(
    "from rdt.transformers.categorical import OneHotEncoder",
    "from sklearn.preprocessing import OneHotEncoder"
)
text = text.replace(
    "ohe = OneHotEncoder()",
    'ohe = OneHotEncoder(sparse_output=False, handle_unknown="ignore")'
)
text = text.replace(
    "ohe.fit(data)",
    "ohe.fit(np.asarray(data).reshape(-1, 1))"
)
text = text.replace(
    "ohe.transform(data)",
    "ohe.transform(np.asarray(data).reshape(-1, 1))"
)
text = text.replace(
    "return encoder.reverse_transform(data)",
    "return encoder.inverse_transform(np.asarray(data)).reshape(-1)"
)
text = text.replace(
    "ohe.output_dimensions",
    "len(ohe.categories_[0])"
)

if "import numpy as np" not in text:
    text = text.replace("import pandas as pd", "import numpy as np\nimport pandas as pd")

path.write_text(text)
print("Transformer patch applied")
```
