```python
from pathlib import Path
import json

import pandas as pd
from sklearn.model_selection import train_test_split


# ============================================================
# Configuration
# ============================================================

INPUT_CSV = Path("Mushroom data.csv")
OUTPUT_DIR = Path("mushroom_processed")

TARGET = "Mushroom_quality"
POSITIVE_LABEL = "p"
RANDOM_STATE = 42

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# Feature descriptions
# ============================================================

DESCRIPTIONS = {
    "Mushroom_quality": (
        "The binary classification target indicating whether the mushroom "
        "is edible or poisonous (e = edible, p = poisonous)."
    ),
    "cap_shape": (
        "The shape of the mushroom cap, such as bell, conical, convex, "
        "flat, knobbed, or sunken."
    ),
    "cap_surface": (
        "The texture of the mushroom cap surface, such as fibrous, "
        "grooved, scaly, or smooth."
    ),
    "cap_color": (
        "The observed color of the mushroom cap."
    ),
    "bruises": (
        "Indicates whether the mushroom has bruises "
        "(t = bruises present, f = no bruises)."
    ),
    "odor": (
        "The odor category of the mushroom, such as almond, anise, "
        "creosote, fishy, foul, musty, none, pungent, or spicy."
    ),
    "gill_attachment": (
        "Describes how the mushroom gills are attached to the stalk."
    ),
    "gill_spacing": (
        "Describes the spacing between the mushroom gills, such as close, "
        "crowded, or distant."
    ),
    "gill_size": (
        "The relative size of the mushroom gills "
        "(b = broad, n = narrow)."
    ),
    "gill_color": (
        "The observed color of the mushroom gills."
    ),
    "stalkshape": (
        "The shape of the mushroom stalk "
        "(e = enlarging, t = tapering)."
    ),
    "stalk_root": (
        "The root structure at the base of the stalk. The value unknown "
        "indicates that the original dataset recorded the feature as missing."
    ),
    "stalk_surface_above_ring": (
        "The texture of the stalk surface above the mushroom ring."
    ),
    "stalk_surface_below_ring": (
        "The texture of the stalk surface below the mushroom ring."
    ),
    "stalk_color_above_ring": (
        "The color of the stalk above the mushroom ring."
    ),
    "stalk_color_below_ring": (
        "The color of the stalk below the mushroom ring."
    ),
    "veil_color": (
        "The observed color of the mushroom veil."
    ),
    "ring_number": (
        "The number of rings present on the mushroom stalk."
    ),
    "ring_type": (
        "The structural type of the ring on the mushroom stalk."
    ),
    "spore_print_color": (
        "The color produced by the mushroom spore print."
    ),
    "population": (
        "The population pattern in which the mushroom grows, such as "
        "abundant, clustered, numerous, scattered, several, or solitary."
    ),
    "habitat": (
        "The habitat where the mushroom was observed, such as grasses, "
        "leaves, meadows, paths, urban areas, waste areas, or woods."
    ),
}


# ============================================================
# Load and clean
# ============================================================

if not INPUT_CSV.exists():
    raise FileNotFoundError(
        f"Could not find input file: {INPUT_CSV.resolve()}"
    )

df = pd.read_csv(INPUT_CSV)

print("Original shape:", df.shape)

df.columns = df.columns.astype(str).str.strip()

for column in df.columns:
    if pd.api.types.is_object_dtype(df[column]):
        df[column] = df[column].astype(str).str.strip()

if TARGET not in df.columns:
    raise ValueError(
        f"Target column '{TARGET}' was not found.\n"
        f"Available columns: {df.columns.tolist()}"
    )

# Replace the Mushroom dataset's missing-category marker.
if "stalk_root" in df.columns:
    df["stalk_root"] = df["stalk_root"].replace(
        {"?": "unknown"}
    )

# Remove constant columns, including veil_type.
constant_columns = [
    column
    for column in df.columns
    if column != TARGET and df[column].nunique(dropna=False) <= 1
]

if constant_columns:
    print("Dropping constant columns:", constant_columns)
    df = df.drop(columns=constant_columns)

# Validate target values.
target_values = set(
    df[TARGET].dropna().astype(str).unique()
)

expected_target_values = {"e", "p"}

if target_values != expected_target_values:
    raise ValueError(
        f"Unexpected target values: {target_values}. "
        f"Expected: {expected_target_values}"
    )

if df.isna().any().any():
    missing_counts = df.isna().sum()
    raise ValueError(
        "Missing values remain:\n"
        f"{missing_counts[missing_counts > 0]}"
    )

print("Cleaned shape:", df.shape)
print("\nTarget counts:")
print(df[TARGET].value_counts())
print("\nTarget proportions:")
print(df[TARGET].value_counts(normalize=True))


# ============================================================
# Stratified 60/20/20 split
# ============================================================

train_df, temporary_df = train_test_split(
    df,
    test_size=0.40,
    random_state=RANDOM_STATE,
    stratify=df[TARGET],
)

val_df, test_df = train_test_split(
    temporary_df,
    test_size=0.50,
    random_state=RANDOM_STATE,
    stratify=temporary_df[TARGET],
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("\nSplit sizes:")
print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

print("\nTrain target counts:")
print(train_df[TARGET].value_counts())

print("\nValidation target counts:")
print(val_df[TARGET].value_counts())

print("\nTest target counts:")
print(test_df[TARGET].value_counts())


# ============================================================
# Save CSV files
# ============================================================

train_path = OUTPUT_DIR / "train.csv"
val_path = OUTPUT_DIR / "val.csv"
test_path = OUTPUT_DIR / "test.csv"

train_df.to_csv(train_path, index=False)
val_df.to_csv(val_path, index=False)
test_df.to_csv(test_path, index=False)


# ============================================================
# Build schema JSON
# ============================================================

schema_columns = []

ordered_columns = [
    TARGET,
    *[column for column in df.columns if column != TARGET],
]

for column in ordered_columns:
    categories = sorted(
        df[column]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    description = DESCRIPTIONS.get(
        column,
        f"Categorical feature named {column}.",
    )

    schema_columns.append(
        {
            "name": column,
            "type": "categorical",
            "i2s": categories,
            "size": len(categories),
            "description": description,
        }
    )

schema = {
    "columns": schema_columns,
    "task": "binary_classification",
    "train_size": len(train_df),
    "val_size": len(val_df),
    "test_size": len(test_df),
    "label": TARGET,
}

schema_path = OUTPUT_DIR / "mushroom.json"

with open(schema_path, "w", encoding="utf-8") as file:
    json.dump(
        schema,
        file,
        indent=2,
        ensure_ascii=False,
    )


# ============================================================
# Validation
# ============================================================

saved_train = pd.read_csv(train_path, dtype=str)
saved_val = pd.read_csv(val_path, dtype=str)
saved_test = pd.read_csv(test_path, dtype=str)

with open(schema_path, "r", encoding="utf-8") as file:
    saved_schema = json.load(file)

schema_names = [
    column["name"]
    for column in saved_schema["columns"]
]

assert len(saved_train) == saved_schema["train_size"]
assert len(saved_val) == saved_schema["val_size"]
assert len(saved_test) == saved_schema["test_size"]

assert saved_schema["label"] == TARGET
assert saved_schema["task"] == "binary_classification"

assert list(saved_train.columns) == list(saved_val.columns)
assert list(saved_train.columns) == list(saved_test.columns)
assert set(schema_names) == set(saved_train.columns)

assert set(saved_train[TARGET].unique()) == {"e", "p"}
assert set(saved_val[TARGET].unique()) == {"e", "p"}
assert set(saved_test[TARGET].unique()) == {"e", "p"}

for column_info in saved_schema["columns"]:
    assert column_info["type"] == "categorical"
    assert column_info["description"].strip()
    assert column_info["size"] == len(column_info["i2s"])

print("\nAll validation checks passed.")

print("\nGenerated files:")
print(train_path.resolve())
print(val_path.resolve())
print(test_path.resolve())
print(schema_path.resolve())

print("\nExperiment configuration:")
print(
    {
        "label": TARGET,
        "pos": POSITIVE_LABEL,
        "task": "poisonous mushroom classification",
    }
)
```
